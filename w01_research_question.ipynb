## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 1. How many pages are declining?
decline_rate = df["trend_direction"].str.lower().eq("down").mean()
print("Declining rate:", round(decline_rate, 3))

# 2. CTR by position tier (shows the CTR cliff)
ctr_by_pos = df.groupby("position_tier")["ctr"].mean().round(3)
print("\nCTR by position tier:")
print(ctr_by_pos)

# 3. Median days since last update (shows staleness)
median_update = df["days_since_last_update"].median()
print("\nMedian days since last update:", median_update)

#The starter dataset shows that around 24% of pages are declining, which means many pages may need refresh attention. CTR also drops sharply as position worsens, showing that ranking changes strongly affect user engagement. Additionally, the median days since last update is high, indicating that many pages are stale. These numbers suggest that deciding which pages to refresh first is a meaningful and valuable problem for the next 7 weeks.
```

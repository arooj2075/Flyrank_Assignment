## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
# 1. Ranked actions + reason codes
#Ranked actions
#My model output is turned into a ranked queue of content actions.
#Each row is a page with an action label and a reason code.
#Main actions
#REFRESH — update or improve the content.
#REVIEW — human review needed, not an automatic change.
#HOLD — no change; the page is stable.
#Reason codes
#REFRESH_OPPORTUNITY — stale and underperforming signals.
#STALE_ONLY — old content but performance is acceptable.
#CTR_UNDERPERFORM — CTR is weak for its position.
#STABLE — signals show no clear issue.
#Ranked queue
#The notebook writes the ranked queue to work/outputs/action_playbook_queue.csv.
#Higher rank = stronger case for action.
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

```python
# 2. Intended use and limits
#Intended use
#This playbook is a decision‑support tool for content teams.
#It suggests which pages to look at first and why.
#It is not a fully automated production system.
#Limits
#The model is directional, not perfect.
#It relies on past signals only and cannot see future events.
#Seasonal content, campaigns, and special pages may behave differently.
#Some pages may be mis‑ranked and need human judgment.
#Scope
#The playbook is meant for prioritization and review, not for automatic edits.
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

```python
# 3. Human review + the no‑go list
#Human review rules
#Every recommended action must be reviewed by a human before changes are made.
#Humans check: business context, content quality, and recent updates.
#No‑go cases (should NOT be automated)
#- Legal, compliance, or policy pages.
#- High‑risk product pages or critical flows.
#- Pages with recent major updates not yet reflected in the data.
#- Any page where the model’s reason code conflicts with known strategy.
#Human role
#The model proposes; humans decide.
#Automation is explicitly out of scope for this playbook.
```

```python
# 4. Monitoring / retrain triggers
#Monitoring
#The playbook should be monitored with simple metrics:
#- How often recommended refreshes are accepted.
#- Whether refreshed pages show directional improvement.
#Retrain triggers
#Retrain or update the model when:
#- The site structure changes significantly.
#- New content types are introduced.
#- Search behavior shifts over time.
#- The model’s recommendations stop matching human judgment.
#Frequency
#This is a light, non‑production setup; retrain can be periodic (e.g., quarterly) or when major changes occur.
```
